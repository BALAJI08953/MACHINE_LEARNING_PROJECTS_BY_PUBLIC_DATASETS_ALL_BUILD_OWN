```python
import numpy as np
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_diabetes
df=load_diabetes()
df
```

```python
my_dataset=pd.DataFrame(df.data)
```

```python
my_dataset
```

```python
my_dataset.columns=df.feature_names
```

```python
my_dataset
```

```python
X=my_dataset
y=df.target
```

```python
#train and test splillting happens here 
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.35,random_state=42)
```

```python
X_train
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
X_train=scaler.fit_transform(X_train)
```

```python
X_test=scaler.fit_transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_val_score
reg=LinearRegression()
mse=cross_val_score(reg,X_train,y_train,scoring='neg_mean_squared_error',cv=5)
```

```python
np.mean(mse)
reg.fit(X_train,y_train)
```

```python
my_test=reg.predict(X_test)
#to print all the my_test
```

```python
import seaborn as sns 
sns.displot(my_test-y_test,kind='kde')
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

print("R2:", r2_score(y_test, my_test))

print("MAE:", mean_absolute_error(y_test, my_test))

print("MSE:", mean_squared_error(y_test, my_test))

print("RMSE:", np.sqrt(mean_squared_error(y_test, my_test)))
```

RIDGE AND LASSO ALGO WHICH ARE USED FOR THR PREVENTION OF OVERFITTING AND REDUCTING THE OR OPTIMAL FEATURE SELECTION


```python
from sklearn.linear_model import Ridge
from sklearn.model_selection import GridSearchCV
my_level =Ridge()
```

```python
my_level
```

```python
params={"alpha":[10,20,3,15,45,78,36,12,40,25,35,65,6,7,8]}
ridcv=GridSearchCV(my_level,params,scoring='neg_mean_squared_error',cv=10)
ridcv.fit(X_train,y_train)
```

```python
print(ridcv.best_params_)
```

```python
print(ridcv.best_score_)
```

```python
ridcv.predict(X_test)
```

```python
ridcv
```

```python
sns.displot(ridcv-y_test,kind='kde')
```

```python
my_score=r2_score(ridcv,y_test)
my_score
```

```python
lasscv=GridSearchCV(my_love,params,scoring='neg_mean_squared_error',cv=10)
lasscv.fit(X_train,y_train)
```
